### syntax for UDF(user defined functions)

CREATE FUNCTION function_name(parameters datatypes) \n
returns datatype \n
return logic

```sql
CREATE OR REPLACE FUNCTION datamaster.bronze.full_name_udf(first_name STRING, last_name STRING) 
RETURNS STRING
RETURN first_name || ' ' || last_name;
```

```sql
SELECT customer_id, datamaster.bronze.full_name_udf(first_name, last_name) as full_name, reviews from  datamaster.bronze.customer_reviews
```

```sql
-- Create a complex UDF for sentiment analysis
CREATE or REPLACE FUNCTION datamaster.bronze.sentiment_analysis(reviews STRING)
RETURNS STRING
RETURN 
    CASE 
        WHEN reviews LIKE '%amazing%' OR reviews LIKE '%great%' OR reviews LIKE '%love%' OR reviews LIKE '%fantastic%' OR reviews LIKE '%recommend%' OR reviews LIKE '%Excellent%' THEN 'Positive'

        WHEN reviews LIKE '%not good%' OR reviews LIKE '%disappointed%' OR reviews LIKE '%terrible%' OR reviews LIKE '%never buying%' 
        THEN 'Negative'

        ELSE 'Neutral'
    END;
```

```sql
SELECT customer_id, datamaster.bronze.full_name_udf(first_name, last_name) as full_name, reviews, datamaster.bronze.sentiment_analysis(reviews) as sentiment from datamaster.bronze.customer_reviews
```
